```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Fashion_Retail_Sales_Cleaned_Transformed.csv')
```

```python
df.shape
```

```python
Total_Revenue = df['Sales_Amount'].sum()
Total_Revenue
```

```python
Total_Profit = df['Profit'].sum()
Total_Profit
```

```python
Total_Orders = df['Order_ID'].count()
Total_Orders
```

```python
Total_Customers = df['Customer_ID'].nunique()
Total_Customers
```

```python
Average_Order_Value = Total_Revenue / Total_Orders
Average_Order_Value
```

```python
Profit_Margin = (Total_Profit / Total_Revenue )*100
Profit_Margin
```

```python
df['Previous_Month_Sales'] = df['Previous_Month_Sales'].fillna(0)
df['Previous_Month_Sales']
```

```python
monthly_summary = df.groupby('Year_Month')['Sales_Amount'].sum().reset_index()

monthly_summary['Previous_Month_Sales'] = (
    monthly_summary['Sales_Amount'].shift(1)
)

df['Previous_Month_Sales'] = df['Previous_Month_Sales'].fillna(0)
df['Previous_Month_Sales']

monthly_summary
```

```python
# Calculate monthly revenue growth percentage
monthly_summary['Growth_%'] = ((monthly_summary['Sales_Amount'] - monthly_summary['Previous_Month_Sales']) / monthly_summary['Previous_Month_Sales']) * 100
 
monthly_summary
monthly_summary.loc[
    monthly_summary['Previous_Month_Sales'].isna(),
    'Growth_%'
] = float('nan')

monthly_summary
```

```python
Growth =  (monthly_summary['Previous_Month_Sales'].sum() / monthly_summary['Sales_Amount'].sum()) * 100
Growth
```

```python
print(f'''KPI CALCULATIONS:

1) Total Revenue              : {Total_Revenue:.2f}
2) Total Profit               : {Total_Profit:.2f}
3) Total Orders               : {Total_Orders}
4) Total Customers            : {Total_Customers}
5) Average Order Value (AOV)  : {Average_Order_Value:.2f}
6) Profit Margin              : {Profit_Margin:.2f} %
''')
```

```python
# Exclude January because its growth rate is NaN
average_growth = monthly_summary['Growth_%'].mean()

median_growth = monthly_summary['Growth_%'].median()

print(f"Average Monthly Growth: {average_growth:.2f}%")
print(f"Median Monthly Growth: {median_growth:.2f}%")
```

## Data Cleaning 

#### Handelling Order Date

```python
from datetime import datetime
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format = "mixed",
                                  dayfirst=True) 
```

```python
df['Order_Date']
```

```python
df['Order_Year'] = df['Order_Date'].dt.year.astype('Int64')
```

```python
df['Order_Month'] = df['Order_Date'].dt.month_name()
```

```python
df['Order_Quarter'] = 'Q' + df['Order_Date'].dt.quarter.astype('string')
```

```python
df["Order_Quarter"] = "Q" + df["Order_Date"].dt.quarter.fillna(0).astype(int).astype(str)
```

```python
df['Order_Quarter']
```

```python
df.shape
```

#### Handelling Units_Sold

```python
df['Units_Sold'] = df['Units_Sold'].where(df['Units_Sold'] > 0 , np.nan)
```

```python
df['Units_Sold'].isna().sum()
```

```python
unit_median = df['Units_Sold'].median()
```

```python
df['Units_Sold'] = df['Units_Sold'].fillna(unit_median)
```

```python
df['Units_Sold'].isna().sum()
```

#### Handelling Unit_Price

```python
df['Unit_Price'].isna().sum()
```

```python
median_price = df.groupby('Product_Name')['Unit_Price'].transform('median')
```

```python
df['Unit_Price'] = df['Unit_Price'].fillna(median_price)
```

```python
df['Unit_Price'].isna().sum()
```

#### Handelling Discount_% and Sales Amount

```python
df['Discount_%'].isna().sum()
```

```python
df['Discount_%'] = df['Discount_%'].where(df['Discount_%'].between(0,100), np.nan)
```

```python
df['Discount_%'].isna().sum()
```

```python
df.loc[df["Sales_Amount"] <= 0, "Sales_Amount"] 
```

```python
df.loc[df["Sales_Amount"] <= 0, ["Sales_Amount", "Discount_%"]]
```

```python
df.loc[df["Sales_Amount"] <= 0, "Sales_Amount"] = np.nan
```

```python
df['Sales_Amount'] = df['Sales_Amount'].fillna(df["Units_Sold"] * df["Unit_Price"] * (1 - df["Discount_%"] / 100))
```

```python
df.loc[[969, 1110, 2323, 2324, 2798, 2812, 3847, 3970],["Units_Sold", "Unit_Price", "Discount_%", "Sales_Amount"]]
```

```python
df["Discount_%"] = df["Discount_%"].fillna(
    (1 - df["Sales_Amount"] / (df["Units_Sold"] * df["Unit_Price"])) * 100
)
```

```python
df['Discount_%'].isna().sum()
```

```python
df.columns
```

#### Handelling Profit

```python
df['Profit_Renew'] = df['Sales_Amount'] - df['Cost_Amount']
```

```python
df = df.drop(columns = ['Profit'])
```

```python
df = df.rename(columns = {'Profit_Renew' : 'Profit'})
```

```python
df['Profit']
```

#### Handelling Profit Margin

```python
df['Profit_Margin_Renew'] = (df['Profit'] / df['Sales_Amount']) * 100
```

```python
df['Profit_Margin_Renew'] = df['Profit_Margin_Renew'].round(2)
```

```python
df[['Profit_Margin_%' , 'Profit_Margin_Renew']]
```

```python
df = df.drop(columns = ['Profit_Margin_%'])
```

```python
df = df.rename(columns = {'Profit_Margin_Renew' : 'Profit_Margin_%'})
```

```python
df['Profit_Margin_%'].isna().sum()
```

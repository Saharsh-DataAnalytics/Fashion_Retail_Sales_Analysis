```python
valid_discounts = [0, 5, 10, 15, 20, 25, 30, 35, 40]

mask = ~df['Discount_%'].apply(
    lambda x: any(abs(x - d) < 0.0001 for d in valid_discounts)
)

df[mask][
    ['Product_Name', 'Unit_Price', 'Units_Sold', 'Discount_%', 'Sales_Amount']
]
```

```python
valid_discounts = [0, 5, 10, 15, 20, 25, 30, 35, 40]
```

```python
invalid_discount = ~df['Discount_%'].between(0, 40) | (df['Discount_%'] % 5 != 0)
```

```python
df.loc[
    invalid_discount & (df['Discount_%'] >= 0),
    'Discount_%'
] = (
    df.loc[
        invalid_discount & (df['Discount_%'] >= 0),
        'Discount_%'
    ] / 5
).round() * 5
```

```python
def get_product_discount(row):
    product_discounts = df.loc[
        (df['Product_Name'] == row['Product_Name']) &
        (~invalid_discount),
        'Discount_%'
    ]
    
    if len(product_discounts) > 0:
        return product_discounts.mode().iloc[0]
    
    return df.loc[~invalid_discount, 'Discount_%'].mode().iloc[0]
```

```python
negative_discount = df['Discount_%'] < 0

df.loc[negative_discount, 'Discount_%'] = df.loc[
    negative_discount
].apply(get_product_discount, axis=1)
```

```python
# 1. Dataset overview
print("========== DATASET OVERVIEW ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Duplicate rows:", df.duplicated().sum())


# 2. Missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum()[df.isnull().sum() > 0])


# 3. Data types
print("\n========== DATA TYPES ==========")
print(df.dtypes)


# 4. Unique values in categorical columns
categorical_cols = [
    'Brand',
    'Product_Category',
    'Product_Name',
    'City',
    'Segment'
]

print("\n========== CATEGORICAL OVERVIEW ==========")

for col in categorical_cols:
    print(f"\n{col}")
    print("Unique values:", df[col].nunique(dropna=False))
    print(df[col].value_counts(dropna=False).head(20))


# 5. Numeric column validation
print("\n========== NUMERIC VALIDATION ==========")

numeric_cols = [
    'Unit_Price',
    'Units_Sold',
    'Discount_%',
    'Sales_Amount',
    'Profit',
    'Profit_Margin_%'
]

print(df[numeric_cols].describe().T)

print("\nNon-positive unit prices:")
print((df['Unit_Price'] <= 0).sum())

print("\nZero or negative quantities:")
print((df['Units_Sold'] <= 0).sum())

print("\nDiscounts outside 0–40:")
print((~df['Discount_%'].between(0, 40)).sum())


# 6. Date validation
print("\n========== DATE VALIDATION ==========")

print("Missing order dates:", df['Order_Date'].isna().sum())

print("\nOrder years inconsistent with Order_Date:")
print(
    (
        df['Order_Year'].notna()
        & df['Order_Date'].notna()
        & (df['Order_Year'] != df['Order_Date'].dt.year)
    ).sum()
)

print("\nOrder months inconsistent with Order_Date:")
print(
    (
        df['Order_Month'].notna()
        & df['Order_Date'].notna()
        & (df['Order_Month'] != df['Order_Date'].dt.month)
    ).sum()
)

print("\nOrder quarters inconsistent with Order_Date:")
print(
    (
        df['Order_Quarter'].notna()
        & df['Order_Date'].notna()
        & (
            df['Order_Quarter']
            != 'Q' + df['Order_Date'].dt.quarter.astype(str)
        )
    ).sum()
)


# 7. Formula consistency
print("\n========== FORMULA VALIDATION ==========")

expected_sales = (
    df['Unit_Price']
    * df['Units_Sold']
    * (1 - df['Discount_%'] / 100)
)

print(
    "Sales_Amount mismatches:",
    (~np.isclose(
        df['Sales_Amount'],
        expected_sales,
        rtol=1e-5,
        atol=0.01,
        equal_nan=True
    )).sum()
)

expected_profit = df['Sales_Amount'] - df['Unit_Price']

print(
    "Profit mismatches:",
    (~np.isclose(
        df['Profit'],
        expected_profit,
        rtol=1e-5,
        atol=0.01,
        equal_nan=True
    )).sum()
)

expected_margin = df['Profit'] / df['Sales_Amount'] * 100

print(
    "Profit margin mismatches:",
    (~np.isclose(
        df['Profit_Margin_%'],
        expected_margin,
        rtol=1e-5,
        atol=0.01,
        equal_nan=True
    )).sum()
)


# 8. Helper columns
print("\n========== HELPER COLUMNS ==========")

helper_cols = ['Typical_Price', 'Price_Ratio']

print("Helper columns still present:")
print([col for col in helper_cols if col in df.columns])
```

```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('Fashion_Retail_Sales_Cleaned_Final.csv')
```

```python
df.columns
```

```python
df[['Sales_Amount','Profit']].describe()
```

```python
q1_sales = 998
q3_sales = 1522.9
q1_profit = 228.105000
q3_profit = 857.385000
```

```python
df['Sales_Category'] = df['Sales_Amount'].apply(
    lambda x : 'Low Sales' if x <= q1_sales
    else 'Medium Sales' if q1_sales < x <= q3_sales
    else 'High Sales'
                                            )
```

```python
df[['Sales_Amount','Sales_Category']]
```

```python
def categorize_profit(x):
    if x < 0:
        return 'Loss'
    elif x == 0:
        return 'Break-Even'
    elif x < q3_profit:
        return 'Low Profit'
    else:
        return 'High Profit'

df['Profit_Category'] = df['Profit'].apply(categorize_profit)
```

```python
df[['Profit' , 'Profit_Category']]
```

```python
Customer_Order_Count = df.groupby('Customer_ID')['Order_ID'].nunique()
df['Customer_Order_Count'] = df['Customer_ID'].map(Customer_Order_Count)
df['Customer_Order_Count']
```

```python
df['Customer_Type'] = df['Customer_Order_Count'].apply(
    lambda x: 'One-Time Customer' if x == 1
    else 'Occasional Customer' if 2 <= x <= 6
    else 'Frequent Customer'
)
```

```python
df[['Customer_Order_Count', 'Customer_Type']]
```

```python
Product_Sales = df.groupby('Product_Name')['Sales_Amount'].sum()
Product_Sales
```

```python
Product_Sales.describe()
```

```python
q1_product_sales = 142089.550
q3_product_sales = 322299.325
```

```python
Product_Performance = Product_Sales.apply(
    lambda x: 'Low Performer' if x <= q1_product_sales
    else 'Average Performer' if x <= q3_product_sales
    else 'High Performer'
)
```

```python
df['Product_Performance'] = df['Product_Name'].map(Product_Performance)
```

```python
df[['Product_Name','Product_Performance' ]].sort_values(by = 'Product_Performance')
```

```python
Order_Units = df.groupby('Order_ID')['Units_Sold'].sum()
```

```python
df['Total_Units_Per_Order'] = df['Order_ID'].map(Order_Units)
```

```python
df[['Order_ID','Total_Units_Per_Order']]
```

```python
df['Total_Units_Per_Order'].describe()
```

```python
df['Order_Size'] = df['Total_Units_Per_Order'].apply(
    lambda x: 'Small Order' if x <= 2
    else 'Medium Order' if x <= 5
    else 'Large Order'
)
```

```python
df[['Total_Units_Per_Order', 'Order_Size']]
```
